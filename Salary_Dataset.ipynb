```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(data)
```

```python
x = df[['YearsExperience']]
y = df[['Salary']]
```

```python
sns.boxplot(x)
plt.title('Year of Experience')
plt.show()

sns.boxplot(y)
plt.title('salary')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.2)
lr = LinearRegression()

lr.fit(x,y)

y_pred = lr.predict(x)

r2_score(y,y_pred)
```

```python
sns.scatterplot(x=x['YearsExperience'],y=y['Salary'])
sns.lineplot(x=x['YearsExperience'], y=y_pred.flatten(), color='red')
```
