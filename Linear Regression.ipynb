```python
X = df.drop(columns=['charges'],axis=1)
y = df['charges']

categorical_feature = X.select_dtypes(include='object').columns
numerical_feature = X.select_dtypes(exclude='object').columns

print("catergorical:",categorical_feature)
print("numberical:",numerical_feature)

```

```python
X = pd.get_dummies(X,columns=['smoker','sex','region'],drop_first=True)
```

```python
X.shape, X.columns
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=42)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```



scaler.fit_transform(X)

X_train, X_test = train_test_split(...)

❌ This is wrong.

Why?

Because the scaler has already seen all the data, including the test set. Information from the test set has "leaked" into training.

The correct order is:

X_train, X_test = train_test_split(...)

scaler.fit(X_train)

X_train = scaler.transform(X_train)

X_test = scaler.transform(X_test)

# Golden Rule:

Never learn anything from the test set.

The test set should remain unseen until it's time to evaluate the model.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

```python
model = LinearRegression()
model = model.fit(X_train_scale,y_train)
y_pred = model.predict(X_test_scale)
```

```python
result = pd.DataFrame({'Actual':y_test,'Prediected':y_pred})
result.head(10)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
coefficients = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": model.coef_
})

coefficients.sort_values(by="Coefficient", ascending=False)
```
